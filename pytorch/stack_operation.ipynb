```python
import torch
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
tensor_1 = torch.tensor([1,2,3])
tensor_2 = torch.tensor([4, 5, 6])

print(tensor_1.ndim)
print(tensor_1.size())
```

```python
row = torch.stack((tensor_1, tensor_2), dim=0)
print(row)

print("----------")

col = torch.stack((tensor_1, tensor_2), dim=1)
print(col)
```

```python
tensor_3 = torch.tensor([7, 8, 9])
```

```python
row = torch.stack((tensor_1, tensor_2, tensor_3), dim=0)
print(row)

print("----------")

col = torch.stack((tensor_1, tensor_2, tensor_3), dim=1)
print(col)
```
