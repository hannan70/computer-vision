### Creating noise image using pytorch

```python

# Image size: Height=256, Width=256, Channels=3 (RGB)
noise_tensor = torch.rand(256, 256, 3)  # Values between 0 and 1

# convert numpy for visualization
noise_image = noise_tensor.numpy()

plt.imshow(noise_image, cmap="gray")
plt.axis("off")
plt.show()
```

#### Tensor zero's and one's

```python
zeros = torch.zeros(10)
print(zeros)
print(type(zeros))
print(zeros.ndim)
```

```python
zeros = torch.zeros(256, 256)
print(zeros)
print(type(zeros))
print(zeros.ndim)
```

```python
zeros_tensor_image = zeros.numpy()
plt.imshow(zeros_tensor_image, cmap="gray", vmin=0, vmax=1)
plt.title("Dark Image")
plt.axis("off")
plt.show()
```

```python
""" 
vmin=0, vmax=1
Sets the value range for mapping pixel intensity.
0 → black
1 → white
"""
# create tensor one's 
tensor_ones = torch.ones(256, 256)

# Convert to numpy for visialization
tensor_ones_image = tensor_ones.numpy()

# plot image
plt.imshow(tensor_ones_image, cmap="gray", vmin=0, vmax=1)
plt.title("White Image")
plt.axis("off")
plt.show()
```

### Tensor Data Types

```python
default_tensor = torch.tensor([1.0, 2.0])
print(default_tensor.dtype)
```

```python
"""
Single precision float
torch.float32 or torch.float
32 bits
"""
float32_data = torch.tensor([1.0, 2.0], dtype=torch.float32)
float_data = torch.tensor([1.0, 2.0], dtype=torch.float)
print(float32_data.dtype)
print(float_data.dtype)

```

```python
"""
* Double precision float
torch.float64 (or torch.double)
64 bits
"""
float64_data = torch.tensor([1.0, 2.0], dtype=torch.float64)
double_data = torch.tensor([1.0, 2.0], dtype=torch.double)
print(float64_data.dtype)
print(double_data.dtype)
```

```python
"""
* Half precision float
torch.float16 (or torch.half)
16 bits
"""
float16_data = torch.tensor([1.0, 2.0], dtype=torch.float16)
half_data = torch.tensor([1.0, 2.0], dtype=torch.half)
print(float16_data.dtype)
print(half_data.dtype)
```

#### Impact of data type of memory

```python
float32_tensor = torch.ones(1000, dtype=torch.float32)
float64_tensor = torch.ones(1000, dtype=torch.float64)

# element_size() --> size of the element in bytes
# nelement() --> total number of element in tensor

print("Size of float32 --> ", float32_tensor.element_size())
print("Total number of element of float32 -->", float32_tensor.nelement())
print("Memory used by float32 tensor : ", float32_tensor.element_size()*float32_tensor.nelement(), "Bytes")

print("==========")

print("Size of float64 --> ", float64_tensor.element_size())
print("Total number of element of float64 -->", float64_tensor.nelement())
print("Memory used by float64 tensor : ", float64_tensor.element_size()*float64_tensor.nelement(), "Bytes")
```
