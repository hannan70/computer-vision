#### Tensor Manipulation

- reshape
- slicing
- joining or splitting
- transposing and permuting dimension

```python
original_tensor = torch.arange(15)
print(original_tensor)
print(original_tensor.nelement())
print(original_tensor.ndim) 
```

```python
reshape_tensor = original_tensor.reshape(3, 5)
print(reshape_tensor.nelement())
print(reshape_tensor.ndim) 
print(reshape_tensor)
```

```python
# all data calculate automatically -1
flattened_tensor = original_tensor.view(-1)
flattened_tensor
print(original_tensor.is_contiguous()) # order are same not change
print(flattened_tensor)
```

### Joining Tensor

```python
tensor_01 = torch.tensor([[1,2,3], 
                          [4,5,6]])
tensor_02 = torch.tensor([[7,8,9], 
                          [10,11,12]])

concat_tensor_row = torch.cat((tensor_01, tensor_02), dim=0)
concat_tensor_col = torch.cat((tensor_01, tensor_02), dim=1)

print(tensor_01)
print(tensor_02)
print("-----")
print(concat_tensor_row)
print(concat_tensor_col)
```

```python
import torch

# Create a tensor with values from 0 to 11
original_tensor = torch.arange(12)

# Split into 5 chunks along dimension 0
# Since 12 is not divisible by 5, the first few chunks will have 3 elements
# and the rest will have 2 elements
chunks = torch.chunk(original_tensor, 5, dim=0)
print(chunks)

# Print each chunk
for chunk in chunks:
    print(chunk)
```

```python
splits = torch.split(original_tensor, 5, dim=0)
print(splits)

# Print each chunk
for split in splits:
    print(split)
```

### Transpose
- transpose() -> swap two dimension. max and min
- permute() -> rearranges all dimension in the specified order

```python
original_tensor = torch.arange(24).reshape(12, 2)

transpose_tensor = original_tensor.transpose(0, 1)


print(original_tensor)

print(original_tensor.shape)
print(transpose_tensor.shape)

print(transpose_tensor)
```

```python
""" 
change the dimension of order

1ম dimension হবে আগের 3য় dimension (index 2)

2য় dimension হবে আগের 1ম dimension (index 0)

3য় dimension হবে আগের 2য় dimension (index 1)

"""
original_tensor = torch.arange(24).reshape(2, 3, 4)



print(original_tensor)

permute_tensor = original_tensor.permute(2, 0, 1)
print(permute_tensor)
```

### Cloning and Detaching

```python
import copy
a = [1,2,3]
b = copy.deepcopy(a)

print(a)
print(b)
```

```python
"""
requires_grad=True
PyTorch-কে বলে দিচ্ছে যে এই tensor-এর উপর যেকোনো অপারেশন ট্র্যাক করো যেন পরে gradient বের করা যায়।

"""
tensor = torch.ones(3, 3, requires_grad=True) # part of computation graph

clone_tensor = tensor.clone() # part of computation graph

print(clone_tensor)
```

```python
detached_tensor = tensor.detach() # this code will detach the tensor from the compution graph
detached_tensor
```
