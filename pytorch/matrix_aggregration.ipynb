```python
import torch
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
matrix = torch.tensor([[1,2,3],
                       [4,5,6]])

```

```python
# total sum
total_sum = torch.sum(matrix)
total_sum
```

```python
# every column sum
col_sum = torch.sum(matrix, dim=0)
col_sum
```

```python
# every row sum
row_sum = torch.sum(matrix, dim=1)
row_sum
```

```python
# every row mean
row_mean = torch.mean(matrix.float(), dim=1)
row_mean
```

```python
# max value
max_val = torch.max(matrix)
max_val
```

```python
row_max = torch.max(matrix, dim=1)
row_max.values # আউটপুট হবে [3, 6] কারণ প্রথম সারির max=3, দ্বিতীয় সারির max=6।
```

```python
row_min = torch.min(matrix, dim=1)
row_min.values # আউটপুট হবে [1, 4] কারণ প্রথম সারির max=1, দ্বিতীয় সারির max=4।
```

```python
# all element Multiplication 
multiplication_all  = torch.prod(matrix)
multiplication_all
```

```python
# col std 
std_col  = torch.std(matrix.float(), dim=0)
std_col
```

```python
# row std 
std_row  = torch.std(matrix.float(), dim=1)
std_row
```

### view and reshape

```python
a = torch.randn(3, 4)
print(a.is_contiguous())
print(a.shape)
# transpose -> could be non- contiguous
b = a.t()
print(b.is_contiguous())
print(b.shape)
```

```python
# view() method works in contiguous data
# it's not copy the main data

# reshape() -> work contiguous and non contiguous both
# copy the main data
```

```python
tensor = torch.arange(12)

reshape_tensor_view = tensor.view(2, 6)
reshape_tensor_reshpae = tensor.reshape(2, 6)

print(reshape_tensor_view)
print(reshape_tensor_reshpae)

print(reshape_tensor_view.shape)
print(reshape_tensor_reshpae.shape)

print(tensor.is_contiguous())
```

```python
tensor3d = torch.arange(24).reshape(2, 3, 4)
tensor3d
```

```python
flattened = tensor3d.view(-1)
flattened
```
