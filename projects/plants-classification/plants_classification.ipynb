```python
results = load_model("/content/plants-classification-3/test/Bayam/aug_0_1216_jpg.rf.0f0023e9fb39f3a67e852ec33ff7de5d.jpg")
```

```python
for res in results:
  res.show()
```

```python
for res in results:
  probs = res.probs

  top1_index = probs.top1

  top1_conf = probs.top1conf

  # get class name
  class_name = res.names[top1_index]

  print(f"Class: {class_name}, Confidence: {top1_conf:.2f}%")
  print("probs ===== ", probs)
  print("Index ===== ", top1_index)
  print("Conf Score ===== ", top1_conf)
  print("class name ====== ", class_name)
  res.show()
```
