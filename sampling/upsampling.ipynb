#### Upsampling is the process of improving the resolution of an image.

##### Common upsampling Techniques
- Nearest Neighbor Interpolation → পাশের পিক্সেল কপি করে বড় করা।
- Bilinear Interpolation → চারপাশের পিক্সেলের গড় ব্যবহার করে smooth ছবি বানানো।
- Bicubic Interpolation → আরও smooth এবং realistic ছবি তৈরি করে।

```python
import cv2 
import torch
import torch.nn.functional as f 
import matplotlib.pyplot as plt
```

```python
image = torch.tensor([
    [1,2,],
    [5,6,],
]).float().unsqueeze(0).unsqueeze(0) # unsqueeze mean add dymention
```

```python
image.shape
```

```python
# Nearest_neighbour upsampling
upsample_nn = f.interpolate(image, scale_factor=2, mode="nearest")
```

```python
upsample_nn
```

```python
upsample_nn.shape
```

```python
upsampling_b = f.interpolate(image, scale_factor=2, mode="bilinear")
```

```python
upsampling_b
```

```python
upsampling_b.shape
```

```python
# Transposed Convolution
conv_transpose = torch.nn.ConvTranspose2d(1, 1, kernel_size=3, stride=2, padding=1)
output_transpose = conv_transpose(image)
```
