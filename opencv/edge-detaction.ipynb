```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('./assets/deer.png')
original_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(original_image)
```

```python
# soba age detaction
sobel_x = cv2.Sobel(image, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(image, cv2.CV_64F, 0, 1, ksize=3)

sobel_combine = cv2.magnitude(sobel_x, sobel_y)

plt.imshow(sobel_combine)
```

```python
# laplacian
laplacian = cv2.Laplacian(image, cv2.CV_64F, ksize=3)
laplacian_abs = cv2.convertScaleAbs(laplacian)
```

```python
plt.imshow(laplacian)
```

```python
plt.imshow(laplacian_abs)
```

```python
canny_edges = cv2.Canny(image, 100, 200)
plt.imshow(canny_edges)
```

```python
title = ["Original Image", "Sobel", "Laplacian", "Laplacine abs", "Canny"]
image_list = [original_image, sobel_combine, laplacian, laplacian_abs, canny_edges]

plt.figure(figsize=(15, 12))
for i in range(len(image_list)):
    plt.subplot(3, 2, i+1)
    plt.imshow(image_list[i])
    plt.title(title[i])
    plt.axis("off")

plt.show()

```
