```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('./assets/deer.png')

if image is not None:
    print("image is loader")
else:
    print("image is not loader")
```

```python
rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(rgb_image)
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
plt.imshow(gray_image)
```

```python
hist, bins = np.histogram(rgb_image.flatten(), 256, [0, 256])

equalization_image = cv2.equalizeHist(gray_image)

equalization_hist, bins = np.histogram(equalization_image.flatten(), 256, [0, 256])


plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.imshow(gray_image)
plt.title("Original Image")

plt.subplot(2, 2, 2)
plt.plot(hist, color="red")
plt.title("Original Image Histogram")

plt.subplot(2, 2, 3)
plt.imshow(equalization_image)
plt.title("Equalization Image")


plt.subplot(2, 2, 4)
plt.plot(equalization_hist)
plt.title("Equalization Image Histogram")
```

```python
"""
LAB = L lightness, AB= color information

YUV
Y = Brightness
U = handle blue color of the image
V = handle red component of the image
"""

yuv_image = cv2.cvtColor(image, cv2.COLOR_BGR2YUV)

yuv_image[:, :, 0] = cv2.equalizeHist(yuv_image[:, :, 0]) # yuv_image[row, col, channel] 0 mean y channel

# convert color yuv to BGR
equalize_color_image = cv2.cvtColor(yuv_image, cv2.COLOR_YUV2BGR)

# Convert Equalize BGR image to RGB
equalize_color_image_rgb = cv2.cvtColor(equalize_color_image, cv2.COLOR_BGR2RGB)


plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.imshow(rgb_image)
plt.title("Original Image")

plt.subplot(2, 2, 2)
plt.imshow(equalize_color_image_rgb)
plt.title("Equalize Image")
```
