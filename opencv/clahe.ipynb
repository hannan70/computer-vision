```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("./assets/deer.png")
plt.imshow(image)
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.imshow(gray_image)
```

```python
equalized_image = cv2.equalizeHist(gray_image)

# clahe 
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(4,4))

clahe_equalized_image = clahe.apply(gray_image)
```

```python
# make plot

plt.figure(figsize=(14, 12))

plt.subplot(1, 3, 1)
plt.imshow(gray_image)
plt.title("Original Image")

plt.subplot(1, 3, 2)
plt.imshow(equalized_image)
plt.title("Equalized Image")


plt.subplot(1, 3, 3)
plt.imshow(clahe_equalized_image)
plt.title("Clahe Equalized Image")  
```

```python
""" 
YUV
Y = Brightness
U = handle blue color of the image
V = handle red component of the image
"""

yuv_image = cv2.cvtColor(image, cv2.COLOR_BGR2YUV)

yuv_image[:,:, 0] = clahe.apply(yuv_image[:,:, 0]) # apply for first channel for brightness mean Y

# Convert YUV to bgr
yuv_image_to_BGR = cv2.cvtColor(yuv_image, cv2.COLOR_YUV2BGR)

# Convert BGR to RGB

yuv_image_to_BGR_to_RGB = cv2.cvtColor(yuv_image_to_BGR, cv2.COLOR_BGR2RGB)


# Original RGB image
rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 8))

plt.subplot(1, 2, 1)
plt.imshow(rgb_image)
plt.title("Original EGB Image")

plt.subplot(1, 2, 2)
plt.imshow(yuv_image_to_BGR_to_RGB)
plt.title("Clahe Equalize Image")
```
