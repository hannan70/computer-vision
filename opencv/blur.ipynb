```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
image = cv2.imread("./assets/giraffe.png")
plt.imshow(image)
```

```python
original_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(original_image)
```

```python
# avg blur
kernel_s = 10
avg_blur = cv2.blur(original_image, (kernel_s, kernel_s))
plt.imshow(avg_blur)
```

```python
kernel_s = 25 # should odd number
gaussian_blur = cv2.GaussianBlur(original_image, (kernel_s, kernel_s), 0)
plt.imshow(gaussian_blur)
```

```python
median_blur = cv2.medianBlur(original_image, 15)
plt.imshow(median_blur)
```

```python
plt.figure(figsize=(10, 8))


plt.subplot(2, 2, 1)
plt.imshow(original_image)
plt.title("Original Image")
plt.axis("off")


plt.subplot(2, 2, 2)
plt.imshow(avg_blur)
plt.title("Avarage Image")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(gaussian_blur)
plt.title("Gaussian Image")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(median_blur)
plt.title("Median Image")
plt.axis("off")
```
