```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('./assets/deer.png')
original_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(original_image)
```

```python
channels = cv2.split(original_image)
colors = ['red', 'green', 'blue']

for channel, color in zip(channels, colors):
    hist = cv2.calcHist([channel], [0], None, [256], [0, 256])
    plt.plot(hist, color=color)

plt.title("Color histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Freequency")
plt.show()
```

```python
image_hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
h, s, v = cv2.split(image_hsv)

h_hist = cv2.calcHist([h], [0], None, [180], [0, 180])
s_hist = cv2.calcHist([s], [0], None, [256], [0, 256])
v_hist = cv2.calcHist([v], [0], None, [256], [0, 256])

fig, ax = plt.subplots(
                    nrows=1, 
                    ncols=3, 
                    figsize=(15, 5)
                )

ax[0].plot(h_hist, color="orange")
ax[0].set_title("Hue Histogram")
ax[0].set_xlabel("Bins")
ax[0].set_ylabel("Frequency")


ax[1].plot(h_hist, color="green")
ax[1].set_title("Saturation Histogram")
ax[1].set_xlabel("Bins")
ax[1].set_ylabel("Frequency")

ax[2].plot(h_hist, color="blue")
ax[2].set_title("Value Histogram")
ax[2].set_xlabel("Bins")
ax[2].set_ylabel("Frequency")

plt.show()
```
